```sql
-- Day 1: Unity Catalog Setup

CREATE CATALOG IF NOT EXISTS energy
COMMENT 'Spanish electricity market and weather lakehouse';
```

```sql
CREATE SCHEMA IF NOT EXISTS energy.bronze
COMMENT 'bronze schema';
```

```sql
CREATE SCHEMA IF NOT EXISTS energy.silver
COMMENT 'silver schema';
```

```sql
CREATE SCHEMA IF NOT EXISTS energy.gold
COMMENT 'gold schema';
```

```sql
CREATE VOLUME IF NOT EXISTS energy.bronze.raw
COMMENT 'raw data json files as they come from the APIs';
```
